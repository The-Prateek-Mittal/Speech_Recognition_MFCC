#### Creating dataset for training the model 

```python
import sounddevice as sd
import librosa
import numpy as np
import pandas as pd 
```

```python
# Defining functions for recording audio and extracting mfcc coefficients 

def record_audio(duration=3, fs=16000):
    print("Recording...")
    audio = sd.rec(int(duration * fs), samplerate=fs, channels=1, dtype='float32')
    sd.wait()
    print("Done.")
    return audio.flatten()

def extract_mfcc_mean(audio, fs=16000, n_mfcc=20):
    mfcc = librosa.feature.mfcc(y=audio, sr=fs, n_mfcc=n_mfcc)
    mfcc_mean = np.mean(mfcc, axis=1)   # mean of each of 13 coeffs
    return mfcc_mean

# Creating data points for Prateek Mittal
person_name = "Prateek Mittal"  
num_samples = 20

speaker_dict = {person_name: []}

for i in range(num_samples):
    print(f"\n[{i+1}/{num_samples}] Starting recording...")
    audio = record_audio()

    mfcc_mean = extract_mfcc_mean(audio)
    speaker_dict[person_name].append(mfcc_mean.tolist())

print("\nFinal Dictionary:")
print(speaker_dict)
```

```python
# Creating data points for Manvith Soma
person_name = "Manvith Soma" 
num_samples = 20

speaker_dict1 = {person_name: []}

for i in range(num_samples):
    print(f"\n[{i+1}/{num_samples}] Starting recording...")
    audio = record_audio()

    mfcc_mean = extract_mfcc_mean(audio)
    speaker_dict1[person_name].append(mfcc_mean.tolist())

print("\nFinal Dictionary:")
print(speaker_dict1)
```

```python
# Creating data points for Sai Laxman
person_name = "Mudapaka Sai Laxman"        
num_samples = 20

speaker_dict2 = {person_name: []}

for i in range(num_samples):
    print(f"\n[{i+1}/{num_samples}] Starting recording...")
    audio = record_audio()

    mfcc_mean = extract_mfcc_mean(audio)
    speaker_dict2[person_name].append(mfcc_mean.tolist())

print("\nFinal Dictionary:")
print(speaker_dict2)
```

```python
# Merging the 3 dictionaries into one dictionary 
speakers_dict = speaker_dict | speaker_dict1 | speaker_dict2
print(speakers_dict)
```

```python
# Converting the dictionary into dataframe to load in the model.
def dict_to_dataframe(speakers_dict):
    rows = []

    for person, mfcc_list in speakers_dict.items():
        for mfcc_vector in mfcc_list:
            mfcc_vector = np.array(mfcc_vector).flatten()  # ensure 1D
            row = {"label": person}

            # Add MFCC values as mfcc_1, mfcc_2, ..., mfcc_n
            for i, val in enumerate(mfcc_vector):
                row[f"mfcc_{i+1}"] = val

            rows.append(row)

    df = pd.DataFrame(rows)
    return df


df = dict_to_dataframe(speakers_dict)
print(df)
```

```python
# Converting dataframe into csv file to save it.
df.to_csv("speaker_mfcc_dataset.csv", index=False)
```

### Training the Model 

```python
import joblib
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.pipeline import Pipeline
import os
import time 
```

```python
# Loading the dataset
df = pd.read_csv("speaker_mfcc_dataset.csv")
print("Dataset Loaded:", df.shape)
print(df.head())

# Splitting the training and testing dataset
X = df.drop("label", axis=1).values
y = df["label"].values

# Converting speaker names into numbers
label_encoder = LabelEncoder()
y_enc = label_encoder.fit_transform(y)


X_train, X_test, y_train, y_test = train_test_split(
    X, y_enc, test_size=0.2, random_state=42, stratify=y_enc
)

# Building ML pipeline

model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", probability=True, class_weight="balanced"))
])

# Training the model

print("\nTraining SVM Model...")
model.fit(X_train, y_train)

# Evaluation

y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)

print("\n=============== RESULTS ===============")
print("Accuracy:", acc)
print("\nClassification Report:\n", classification_report(
    y_test, y_pred, target_names=label_encoder.classes_
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# Saving the model and label encoder

os.makedirs("saved_model", exist_ok=True)
joblib.dump(model, "saved_model/speaker_svm_model.joblib")
joblib.dump(label_encoder, "saved_model/label_encoder.joblib")

print("\nModel saved to: saved_model/speaker_svm_model.joblib")
print("Label Encoder saved to: saved_model/label_encoder.joblib")
print("======================================")
```

### Real-Time and One-Time Predictions

```python

DURATION = 3             # seconds to record per sample
FS = 16000               # sample rate
N_MFCC = 20             # number of mfcc coefficients

MODEL_PATH = "saved_model/speaker_svm_model.joblib"
LABELENC_PATH = "saved_model/label_encoder.joblib"

# Loading Saved Model and Encoder

print("Loading model...")
model = joblib.load(MODEL_PATH)
label_encoder = joblib.load(LABELENC_PATH)
print("Model & Label Encoder Loaded Successfully!\n")


# Function: Extract MFCC Mean for prediction

def extract_mfcc_mean(audio, sr=16000):
    mfcc = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=N_MFCC)
    mfcc_mean = np.mean(mfcc, axis=1)
    return mfcc_mean


# Real-Time Prediction

def predict_speaker_from_audio(audio):
    mfcc_mean = extract_mfcc_mean(audio)
    X = mfcc_mean.reshape(1, -1)

    probs = model.predict_proba(X)[0]
    idx = np.argmax(probs)
    speaker = label_encoder.inverse_transform([idx])[0]
    confidence = probs[idx]

    return speaker, confidence


# Main Real-Time Loop

print("Real-time Speaker Recognition Started!")
print("Speak for {} seconds when prompted.".format(DURATION))
print("Press CTRL + C to stop.\n")

try:
    while True:
        print("Listening...")
        
        # Record audio
        audio = sd.rec(int(DURATION * FS), samplerate=FS, channels=1, dtype='float32')
        sd.wait()

        audio = audio.flatten()

        # Make prediction
        speaker, conf = predict_speaker_from_audio(audio)

        # Show result
        print("\n--------------------------------------")
        print("Predicted Speaker:", speaker)
        print("Confidence:", round(conf, 3))
        print("--------------------------------------\n")

        time.sleep(1)

except KeyboardInterrupt:
    print("\nReal-Time Recognition Stopped.")
```
